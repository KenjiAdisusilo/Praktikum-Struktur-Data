### Case 1: Coin Change

```python
def coin_change_greedy (amount, coins):
  coins.sort(reverse=True) # Mengurutkan koin dari terbesar ke terkecil agar program selalu mengambil koin terbesar dulu.
  result = [] # Membuat list kosong untuk menyimpan koin yang digunakan.
  for coin in coins: # untuk setiap denomination koin, ambil koin tersebut berulang kali selama memungkinkan.
    while amount >= coin:
      amount -= coin
      result.append (coin)
  return result
# Contoh penggunaan
amount = 57
coins = [25, 10, 5, 1]
change = coin_change_greedy (amount, coins)
print ("Koin yang digunakan:", change)
```

### Case 2: Fractional Knapsack

```python
def fractional_knapsack (items, capacity):
  # items = list of tuples (value, Wieght)
  items = sorted(items, key=lambda x: x[0]/x [1], reverse=True) # Mengurutkan item berdasarkan value-to-weight ratio dari terbesar ke terkecil. inti dari algoritma greedy.
  total_value = 0.0 # Inisialisasi total nilai yang diperoleh
  for value, weight in items:
    if capacity >= weight: # Jika kapasitas cukup → ambil seluruh item
      capacity -= weight
      total_value += value
    else: # Jika kapasitas tidak cukup → ambil sebagian(item), lalu berhenti
      total_value += value * (capacity / weight)
      break
  return total_value

# Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)] # (nilai, berat)
capacity = 50
max_value = fractional_knapsack (items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
```

### Case 3 : Minimum Spanning Tree: Prim dan Kruskal

#### Kode 1

```python
import heapq

def prim_mst (graph, start):
  visited = set ()
  min_heap = [(0, start)]
  total_weight = 0

  while min_heap:
    weight, node = heapq.heappop (min_heap)
    if node in visited:
      continue
    visited.add(node)
    total_weight += weight
    for neighbor, edge_weight in graph[node]:
      if neighbor not in visited:
        heapq.heappush (min_heap, (edge_weight, neighbor))
  return total_weight

# Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
print(f"Total bobot MST (Prim):", prim_mst (graph, 'A'))
```

#### Kode 2

```python
def kruskal_mst (edges, n_nodes) :
  parent = {i: i for i in range (n_nodes)}

  def find(x):
    while parent[x] != x:
      x = parent[x]
    return x
  def union (x, y):
    root_x = find(x)
    root_y = find(y)
    if root_x != root_y:
      parent[root_y] = root_x
      return True
    return False

  edges.sort (key=lambda x: x [2])
  total_weight = 0
  for u, v, weight in edges:
    if union(u, v):
      total_weight += weight
  return total_weight
# edges: (nodel, node2, weight), nodes are represented by integers
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print ("Total bobot MST (Kruskal):", kruskal_mst (edges, 4))
```
