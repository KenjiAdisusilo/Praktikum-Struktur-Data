```python
# Latihan Personal
def count_neighbors(graph):
    result = {}
    for node in graph:
        # Menghitung jumlah elemen dalam list/set tetangga untuk simpul tersebut
        result[node] = len(graph[node])
    return result

#--- Contoh Penggunaan ---
my_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A'],
    'D': ['B']
}
print(count_neighbors(my_graph))
```

```python
from collections import deque

# 1. Representasi Graf Baru (5 Simpul, 7 Sisi)
graph_baru = {
    'V': ['W', 'X'],
    'W': ['V', 'X', 'Y'],
    'X': ['V', 'W', 'Y', 'Z'],
    'Y': ['W', 'X', 'Z'],
    'Z': ['X', 'Y']
}

# Fungsi untuk menampilkan graf [cite: 157]
def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")

# 2. Implementasi BFS [cite: 170]
def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# 2. Implementasi DFS [cite: 191]
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

# --- EKSEKUSI PROGRAM ---
print("=== Tampilan Graf Tak Berarah (5 Simpul, 7 Sisi) ===")
print_graph(graph_baru)

print("\n=== Hasil Traversal BFS dari Simpul 'V' ===")
bfs(graph_baru, 'V')

print("\n\n=== Hasil Traversal DFS dari Simpul 'V' ===")
dfs(graph_baru, 'V')
```

```python
# 3. Modifikasi fungsi BFS Mengembalikan
def bfs_return_list(graph, start):
    visited = set()
    queue = deque([start])
    order = [] # List untuk menyimpan urutan kunjungan

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            order.append(vertex) # Memasukkan simpul ke dalam list
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

    return order # Mengembalikan list hasil urutan

# --- Contoh Pemanggilan ---
print(bfs_return_list(graph_baru, 'V'))
```

```python
# 4. Fungsi find_path Menggunakan DFS
def find_path(graph, start, end, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path = path + [start] # Tambahkan simpul saat ini ke rute jalur

    # Jika sudah sampai di tujuan, kembalikan jalurnya
    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in visited:
            new_path = find_path(graph, neighbor, end, visited, path)
            if new_path: # Jika jalur ke tujuan ditemukan di cabang ini
                return new_path

    return None # Jika tidak ada jalur

# Contoh Pemanggilan mencari jalur dari V ke Z:
print(find_path(graph_baru, 'V', 'Z'))
```

```python
# Fungsi is_connected Menggunakan DFS
def is_connected(graph):
    if not graph:
        return True

    # Ambil salah satu simpul acak sebagai titik mulai
    start_node = list(graph.keys())[0]

    # Lakukan DFS internal untuk mendata semua simpul yang bisa dijangkau
    visited = set()
    def dfs_internal(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_internal(neighbor)

    dfs_internal(start_node)

    # Jika jumlah simpul yang dikunjungi sama dengan total simpul di graf, berarti terhubung
    return len(visited) == len(graph)

# Contoh Pemanggilan:
print(is_connected(graph_baru))
```
