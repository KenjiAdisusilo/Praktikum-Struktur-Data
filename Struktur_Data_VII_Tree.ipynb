Binary Tree(manual)

```python
class Node:
  def __init__(self, data):
    self.data = data # Nilai/data pada node
    self.left = None # Pointer ke left child
    self.right = None # Pointer ke right child
# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
# Fungsi inorder traversal
def inorder(node):
  if node: # Jika node tidak None
    inorder(node.left) # 1. kunjungi kiri
    print(node.data, end=' ') # 2. Print data
    inorder(node.right) # 3. Kunjungi kanan
print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

Binary Search Tree

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None # Root tree awalnya kosong
  def insert(self, root, data):
    if root is None: # Jika kosong :
      return Node(data) # Buat node baru
    if data < root.data: # Jika data < Root:
      root.left = self.insert(root.left, data) # Rekursi ke kiri
    else: # Jika Data >= Root :
      root.right = self.insert(root.right, data) # Rekursi ke kanan
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)
```

Traversal pada BST

```python
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)
def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)
def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

```python
#--Binary Search Tree--#
# Class Node (Sturktur dasar tree)
class Node:
  def __init__(self, data):
    self.data = data # Nilai/data pada node
    self.left = None # Pointer ke anak kiri
    self.right = None # Pointer ke anak kanan

# Class Binary search
class BinarySearchTree:
  def __init__(self):
    self.root = None # Root awalnya kosong
  # Menyisipkan data dengan aturan BST
  def insert(self, root, data):
    if root is None: # Jika node Kosong → Buat node baru
      return Node(data)
    if data < root.data: # Jika data < node sekarang → rekursi ke kiri
      root.left = self.insert(root.left, data)
    else: # # Jika data >= node sekarang → rekursi ke kanan
      root.right = self.insert(root.right, data)
    return root # Kembalikan root

# Pembuatan BST dengan data identitas
bst = BinarySearchTree()
root = None # Mulai dari root kosong
data_list = [14, 30, 9, 75, 101, 7, 21] # [Digit terakhir npm(014, 0 didepan akan eror), tanggal lahir, bulan lahir, ASCII K, ASCII e, Nomor bebas, Nomor bebas]
for value in data_list:
  root = bst.insert(root, value) # insert satu per satu, update root

#---Traversal---#
# Inorder (Left→Root→Right)
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)
# Preorder (Root→Left→Right)
def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)
# Postorder (Left→Right→Root)
def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

print("=== STRUKTUR BINARY SEARCH TREE BERDASARKAN IDENTITAS ===")
print("--Inorder Traversal--")
inorder(root)
print("\n--Preorder Traversal--")
preorder(root)
print("\n--Postorder Traversal--")
postorder(root)


#--Fitur Pencarian BST--#
def search(node, key):
  if node is None or node.data == key:
    return node
  if key < node.data:
    return search(node.left, key)
  return search(node.right, key)

# Uji pencarian
print("\n\n=====Pencarian dalam BST=====")
key = 14 # Test: Digit terakhir NPM
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
key = 50 # Test: Angka yg tidak dalam list
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
```
